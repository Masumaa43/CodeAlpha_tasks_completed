```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

#Load the dataset
df = pd.read_csv("../dataset/credit_risk_dataset.csv")

#Display the first five rows
df.head()
```

```python
#check the number of rows and columns in the dataset
print("Dataset Shape:", df.shape)

#Display column names
print("\nColumn Names:")
print(df.columns.tolist())

#check data types and missing values
print("\nDataset Information:")
print(df.info())

#check missing values in the dataset
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print(df['loan_status'].value_counts())

print("\nTarget distribution:")
print(df['loan_status'].value_counts(normalize=True) * 100)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=df, x='loan_status')
plt.title('Loan Default Distribution')
plt.xlabel('Loan Status (0 = No Default, 1 = Default)')
plt.ylabel('Number of Applicants')
plt.show()
```

```python
df.describe()
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["person_income"], bins=50, kde=True)

plt.title("Distribution of Applicant Income")
plt.xlabel("Annual Income")
plt.ylabel("Number of Applicants")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["loan_amnt"], bins=30, kde=True)

plt.title("Distribution of Loan Amounts")
plt.xlabel("Loan Amount")
plt.ylabel("Number of Applicants")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="loan_status", y="person_income")

plt.title("Income vs Loan Default Status")
plt.xlabel("Loan Status (0 = No Default, 1 = Default)")
plt.ylabel("Annual Income")

plt.show()
```
